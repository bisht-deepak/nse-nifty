```python
SYMBOL_TO_COMPANY= {"IDFC": "Infrastructure Development Finance Co. Ltd.", "INFY": "Infosys Ltd.", "ITC": "ITC Ltd.", "SAIL": "Steel Authority of India Ltd."}
```

```python
COMPANY_TO_SYMBOL= {"Bajaj Auto Ltd.": "BAJAJAUTO", "Eternal Ltd.": "ETERNAL", 
 "Infosys Technologies Ltd.": "INFY", "Reliance Infrastructure Ltd.": "RELINFRA",
 "Tata Communications Ltd.": "TATACOMM", "Vedanta Ltd.": "VEDL"}
```

```python
COMPANY_NAME_STANDARDIZATION = {
    "Infosys Technologies Ltd.": "Infosys Ltd.",
    "Hero Honda Motors Ltd.": "Hero MotoCorp Ltd.",
    "Sterlite Industries (India) Ltd.": "Vedanta Ltd.",
    "Sesa Goa Ltd.": "Vedanta Ltd.",
    "IT C Ltd.": "ITC Ltd.",
}
```

```python
nifty50= (
    pd.read_parquet(
        "../data/processed/nifty50_history.parquet",
        # parse_dates=["Date"],
    )
    .assign(
            symbol= lambda df_: df_["company_name"].map(COMPANY_TO_SYMBOL).fillna(df_["symbol"]),
            company_name= lambda df_: (
                df_["symbol"]
                .map(SYMBOL_TO_COMPANY).fillna(df_["company_name"])
                # .map(COMPANY_NAME_STANDARDIZATION).fillna(df_["company_name"])
            ),
            industry_= lambda df_: (
                df_["industry"]
                .astype(str)
                .str.strip()
                .str.replace(r"\s+", " ", regex=True)
                .map(INDUSTRY_STANDARDIZATION)
            )
    )
    # .assign(
    #     company_name= lambda df_: (
    #         df_["company_name"].map(COMPANY_NAME_STANDARDIZATION).fillna(df_["company_name"])
    #     )
    # )
)

nifty50.head()
```
